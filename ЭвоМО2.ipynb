### **Примеры задач** для эволюционных методов:
1. Настройка коэффициентов по гладкой поверхности (лучше через производную)
2. Маршрут по городам (лучше через структуру с перебором)
3. Долгое моделирование (стоимость оценки)
4. Обучение параметры по данным с шумом (можно ли бороться с шумом)

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

fig, ax = plt.subplots(figsize=(11, 10))
ax.set_xlim(-1, 11)
ax.set_ylim(-4, 4)
ax.axis('off')
ax.set_aspect('equal')

def draw_node(ax, x, y, text, w, h, color, fontsize=11, fontweight='bold'):
    box = FancyBboxPatch(
        (x - w/2, y - h/2), w, h,
        boxstyle="round,pad=0.08,rounding_size=0.15",
        facecolor=color, edgecolor="white", linewidth=2.2, zorder=3
    )
    ax.add_patch(box)
    ax.text(x, y, text, ha='center', va='center', fontsize=fontsize,
            fontweight=fontweight, color='white', zorder=4)

def draw_edge(ax, start, end, color="#888888"):
    arrow = FancyArrowPatch(
        start, end,
        connectionstyle="arc3,rad=0.15",
        arrowstyle="-", linewidth=2, color=color, zorder=2
    )
    ax.add_patch(arrow)

# --- Корень ---
root_pos = (0.8, 0)
draw_node(ax, *root_pos, "Методы\nоптимизации", 2.6, 1.5, "#4C72B0", fontsize=13)

# --- Уровень 1: три ветви ---
level1 = {
    "Аналитические": (4.5, 2.6, "#DD8452"),
    "Траекторные\n(напр. градиентные)": (4.7, 0, "#55A868"),
    "Популяционные": (4.5, -2.6, "#C44E52"),
}

for text, (x, y, color) in level1.items():
    draw_edge(ax, (root_pos[0] + 1.3, root_pos[1]), (x - 1.3, y))
    draw_node(ax, x, y, text, 2.8, 1.3, color, fontsize=11)

pop_pos = level1["Популяционные"][:2]

# --- Уровень 2: ветви от "Популяционные" ---
level2 = {
    "Эволюционные": (8.5, -1.4, "#8172B2"),
    "Роевые": (8.5, -3.8, "#8172B2"),
}

for text, (x, y, color) in level2.items():
    draw_edge(ax, (pop_pos[0] + 1.4, pop_pos[1]), (x - 1.3, y))
    draw_node(ax, x, y, text, 2.4, 1.1, color, fontsize=10.5)

plt.title("Классификация методов оптимизации", fontsize=15, fontweight='bold', pad=15)
plt.tight_layout()
plt.show()
```
